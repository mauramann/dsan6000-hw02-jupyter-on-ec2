```python
import boto3

# Create s3 client, bucket
s3 = boto3.client('s3', region_name='us-east-1')
s3.create_bucket(Bucket='dsan6000-mm5514')
```

```python
import os

parq_files = 'data/'
bucket_name = 'dsan6000-mm5514'
folder_name = 'wikipedia-hourly' 

# Loop through local directory data/
for root, dirs, files in os.walk(parq_files):
    for filename in files:
        local_path = os.path.join(root, filename)
        relative_path = os.path.relpath(local_path, parq_files)
        # Create s3 path, upload file to bucket
        s3_key = f"{folder_name}/{relative_path}".replace("\\", "/") if folder_name else relative_path.replace("\\", "/")
        s3.upload_file(local_path, bucket_name, s3_key)
        # Record successful upload
        print(f"Uploaded {local_path} -> s3://{bucket_name}/{s3_key}")
```
