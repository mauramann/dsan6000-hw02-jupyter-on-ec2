```python
import boto3
from botocore import UNSIGNED
from botocore.client import Config

s3 = boto3.client("s3", config=Config(signature_version=UNSIGNED))

for file in s3.list_objects_v2(Bucket="dsan6000-wikipedia", Prefix="hourly/")["Contents"]:
    name = file["Key"].split("/")[-1]
    if name:
        s3.download_file("dsan6000-wikipedia", file["Key"], f"data/{name}")
```
